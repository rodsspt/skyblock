### Preparação do Dataset: LLM4Binary/decompile-ghidra-100k
Este dataset contém pares de código assembly/binário e código C decompilado. Vamos carregá-lo usando a biblioteca `datasets`.

```python
from datasets import load_dataset

# Carregando o dataset
dataset = load_dataset("LLM4Binary/decompile-ghidra-100k", split = "train")

# Visualizando as colunas disponíveis
print(dataset.column_names)
display(dataset.to_pandas().head())
```

Agora, vamos formatar o dataset para o Unsloth. Geralmente, usamos o `standardize_sharegpt` ou aplicamos um template de chat. Abaixo está um exemplo de como mapear os campos `input` (código assembly) e `output` (código C).

```python
def formatting_prompts_func(examples):
    inputs  = examples["ghidra_asm"]
    outputs = examples["c_code"]
    texts = []
    for input_text, output_text in zip(inputs, outputs):
        # Exemplo de prompt para decompilação
        text = f"### Instruction:\nDecompile the following assembly code into C.\n\n### Input:\n{input_text}\n\n### Response:\n{output_text}"
        texts.append(text)
    return { "text" : texts, }

pass_dataset = dataset.map(formatting_prompts_func, batched = True)
print("Exemplo do prompt formatado:")
print(pass_dataset["text"][0])
```
